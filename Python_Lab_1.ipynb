```python
model_name = "Logistic Regression"
accuracy = 0.87345

# Old, clunky way (avoid this):
print("Accuracy of " + model_name + " is " + str(accuracy))

# f-string way (use this):
print(f"Accuracy of {model_name} is {accuracy}")

# ':.3f' inside the braces rounds a float to 3 decimal places -
# you will use this in almost every experiment to print clean metrics.
print(f"Accuracy of {model_name} is {accuracy:.3f}")
```

```python
values = [12, 45, 67, 23, 89, 34]

print("\nElements in List:")
for i in values:
    print(i)

total = 0
for i in values:
    total += i

average = total / len(values)
print("Average:", average)

threshold = 40
print("Values Greater Than", threshold, ":")
for i in values:
    if i > threshold:
        print(i)

maximum = values[0]
minimum = values[0]

for i in values:
    if i > maximum:
        maximum = i
    if i < minimum:
        minimum = i

print("Maximum Value:", maximum)
print("Minimum Value:", minimum)

count = 0

for i in values:
    if i > 50:
        count += 1

print("Count of Values Greater Than 50:", count)
```
